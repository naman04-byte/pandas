```python
import pandas as pd
```

```python
#to_csv
df=pd.read_csv('deliveries.csv')
```

```python
df.head()
```

```python
temp_df=df.groupby('batsman')['batsman_runs'].sum().reset_index()
```

```python
temp_df.to_csv('batsman_run.csv',index=False)
```

```python
df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum').to_csv('batsman_vs_team.csv')
```

```python
#to_excel
temp_df=df.groupby('batsman')['batsman_runs'].sum()
```

```python
temp_df.to_excel('batsman_runs.xlsx')
```

```python
temp_df2=df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum')
```

```python
with pd.ExcelWriter('output.xlsx') as writer:
    temp_df.to_excel(writer,sheet_name='batsman run')
    temp_df2.to_excel(writer,sheet_name='batsman_vs_team')
```

```python
# to_html
df.query('batsman_runs==6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count').to_html('sixes_heatmap.html')
```

```python
# to_json
df.groupby(['batting_team','batsman'])['batsman_runs'].sum().unstack().to_json('ipl.json')
```

```python
import pymysql
from sqlalchemy import create_engine
```

```python
engine=create_engine("mysql+pymysql://root:@localhost/ipl_deliveries")
```

```python
temp_df.to_sql('batsman_runs',con=engine,if_exists='append')
```

```python
six_df=df.query('batsman_runs==6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count')
six_df.to_sql('six_heatmap',con=engine,if_exists='append')
```
